*Here in the above dataset there is some features which are not required for the prediction so we will drop those features and keep only the required features for the prediction*

```python
# delete college_ID column 
df = df.drop('College_ID', axis=1)
```

```python
df.head()
```

```python
# dummy encoding the categorical features alternative of one hot encoding
from sklearn.preprocessing import LabelEncoder 
le = LabelEncoder()
df['Placement'] = le.fit_transform(df['Placement']) # this convert the categories into numnerical yes = 1 and no = 0
```

```python
for col in df.select_dtypes(include='object').columns:
    dummies = pd.get_dummies(df[col], prefix=col, drop_first=True).astype(int)
```

```python
df.head()
```

```python
dummies.head() # only the interenship coulumn is categorical its made dummy variable for it 
```

```python
# removing the original categorical column and adding the dummy variable column to the dataset
df = df.drop('Internship_Experience', axis=1)
# add dummy col to dataframe 
df = pd.concat([df, dummies], axis=1)
```

```python
df.tail()
```

```python
df.head() # now the dataste is ready for the model 
```

- splitting the dataset 

```python
x = df.drop('Placement', axis=1) # dependent variable 
y = df['Placement'] # independent variable
```

```python
x.head()
```

```python
# importing liabraries for model building
from sklearn.model_selection import train_test_split 
from sklearn.preprocessing import StandardScaler 
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, matthews_corrcoef
```

```python
# splitting the dataset
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42) 
```

- Standardizing the dataset

```python
scaler = StandardScaler() 
x_train_scale = scaler.fit_transform(x_train)
x_test_scale = scaler.transform(x_test)
```

- *Building the model*

**XGBoost Classifier**

```python
from xgboost import XGBClassifier 
# here is model with regularization parameters because model is making more importance to the specfic features
xgb = XGBClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    min_child_weight=5,
    subsample=0.8,
    colsample_bytree=0.8,
    reg_alpha=0.1,
    reg_lambda=1.0,
    random_state=42
)
xgb.fit(x_train, y_train)

y_pred_train = xgb.predict(x_train)
y_pred_test = xgb.predict(x_test)

# for training data 
print('For Training')
print('Classification Report:- \n',classification_report(y_train, y_pred_train))
print("confusion_matrix:- ", confusion_matrix(y_train, y_pred_train))
print("accuracy_score:- ",accuracy_score(y_train, y_pred_train))
print('Mathews correlation coefficient:- ',matthews_corrcoef(y_train, y_pred_train))

# for testing data 
print('For Testing')
print('classification_report:- \n',classification_report(y_test, y_pred_test))
print('confusion_matrix:- ', confusion_matrix(y_test, y_pred_test))
print('accuracy_score:- ',accuracy_score(y_test, y_pred_test))
print('Mathews correlation coefficient:- ',matthews_corrcoef(y_test, y_pred_test))
```

```python
# showing the feature importance of the model
print(xgb.feature_importances_)
```

**Logistic Regression**

```python
from sklearn.linear_model import LogisticRegression 

lg = LogisticRegression()
lg.fit(x_train_scale, y_train)

y_pred_train = lg.predict(x_train_scale)
y_pred_test = lg.predict(x_test_scale)

# for training data 
print('For Training')
print('Classification Report:- \n',classification_report(y_train, y_pred_train))
print("confusion_matrix:- ", confusion_matrix(y_train, y_pred_train))
print("accuracy_score:- ",accuracy_score(y_train, y_pred_train))

# for testing data 
print('For Testing')
print('classification_report:- \n',classification_report(y_test, y_pred_test))
print('confusion_matrix:- ', confusion_matrix(y_test, y_pred_test))
print('accuracy_score:- ',accuracy_score(y_test, y_pred_test))
```

**Decision Tree Classifier**

```python
from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier(min_samples_split=3, min_samples_leaf=4 ,random_state=42)
dt.fit(x_train, y_train)

y_pred_train = dt.predict(x_train)
y_pred_test= dt.predict(x_test)

# for training data 
print('For Training')
print('Classification Report:- \n',classification_report(y_train, y_pred_train))
print("confusion_matrix:- ", confusion_matrix(y_train, y_pred_train))
print("accuracy_score:- ",accuracy_score(y_train, y_pred_train))

# for testing data 
print('For Testing')
print('classification_report:- \n',classification_report(y_test, y_pred_test))
print('confusion_matrix:- ', confusion_matrix(y_test, y_pred_test))
print('accuracy_score:- ',accuracy_score(y_test, y_pred_test))
```

**Random Forest Classifier**

```python
# Random Forest is the best model for this dataset as it gives the best accuracy and feature importance
from sklearn.ensemble import RandomForestClassifier 
import joblib 

rf = RandomForestClassifier(random_state=42)
rf.fit(x_train, y_train)

# Saving the model 
joblib.dump(rf, 'placement_predictor.joblib')

y_pred_train = rf.predict(x_train)
y_pred_test = rf.predict(x_test)

# for training data 
print('For Training')
print('Classification Report:- \n',classification_report(y_train, y_pred_train))
print("confusion_matrix:- ", confusion_matrix(y_train, y_pred_train))
print("accuracy_score:- ",accuracy_score(y_train, y_pred_train))
print('Mathews correlation coefficient:- ',matthews_corrcoef(y_train, y_pred_train))

# for testing data 
print('For Testing')
print('classification_report:- \n',classification_report(y_test, y_pred_test))
print('confusion_matrix:- ', confusion_matrix(y_test, y_pred_test))
print('accuracy_score:- ',accuracy_score(y_test, y_pred_test))
print('Mathews correlation coefficient:- ',matthews_corrcoef(y_test, y_pred_test))
```

### making new dataset with balanced classes 

```python
yes = df[df['Placement'] == 1] # placed studeents 
```

```python
yes.shape
```

```python
yes.head()
```

```python
no = df[df['Placement'] == 0] # not placed students
no_back = no.copy() # copy of no dataset 
```

```python
no.shape
```

```python
no.head()
```

```python
n = no[:1000] # taking last 1000 rows of the not placed students dataset  
n_x = n.drop('Placement', axis=1) # independent variable
n_y = n['Placement'] # dependent variable
```

```python
no = no.sample(n=2000, random_state=42) # taking a sample of 2000 students 'not placed' to balance the dataset 
```

```python
no.shape
```

```python
data = pd.concat([yes, no], axis=0) # combining the placed and not placed students to create a balanced dataset
```

```python
data.shape
```

```python
# shuffle the dataset 
data = data.sample(frac=1, random_state=42).reset_index(drop=True) # shuffling the data and reset index from 0 to n-1
```

```python
data.tail(15) # now the dataset is balanced with equal number of placed and not placed students. This will help the model to learn better and give better accuracy.
```

```python
#data = data.drop('Communication_Skills', axis=1) # dropping the communication skills column as it is not important for the model and it is also a categorical variable which is not suitable for the model.
X = data.drop('Placement', axis=1) # dependent variable 
Y = data['Placement'] # independent variable
```

```python
x_tr, x_te, y_tr, y_te = train_test_split(X, Y, test_size=0.2, random_state=42, shuffle=True, stratify=Y) 
# stratify = Y means that split the data in such a way that the division of the classes in the training and testing data is same as that of the original dataset.
```

```python
# logistic regression model 
from sklearn.pipeline import Pipeline # for pipeline model building
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
model = Pipeline([
    ("polynomial", PolynomialFeatures(degree=2, include_bias=False)),
    ("scaler", StandardScaler()),
    ("classifier", LogisticRegression(max_iter=1000))
])
model.fit(x_tr, y_tr)

y_pred_train = model.predict(x_tr)
y_pred_test= model.predict(x_te)

# for training data 
print('For Training')
print('Classification Report:- \n',classification_report(y_tr, y_pred_train))
print("confusion_matrix:- ", confusion_matrix(y_tr, y_pred_train))
print("accuracy_score:- ",accuracy_score(y_tr, y_pred_train))

# for testing data 
print('For Testing')
print('classification_report:- \n',classification_report(y_te, y_pred_test))
print('confusion_matrix:- ', confusion_matrix(y_te, y_pred_test))
print('accuracy_score:- ',accuracy_score(y_te, y_pred_test))
```

```python
from sklearn.inspection import permutation_importance

result = permutation_importance(
    model, x_te, y_te,
    n_repeats=10, random_state=42
)

for i in range(x_te.shape[1]):
    print(x_te.columns[i], result.importances_mean[i])
```
